```python
from transformers import pipeline
from langdetect import detect
```

```python
bart_summarizer = pipeline(
    "summarization",
    model="kochi-metro24/bart-metro-summarizer"
)

mbart_summarizer = pipeline(
    "summarization",
    model="facebook/mbart-large-50-many-to-many-mmt"
)
```

```python
def hybrid_summarize(text):
    lang = detect(text)

    if lang == "en":
        summary = bart_summarizer(
            text, max_length=120, min_length=30, do_sample=False
        )[0]['summary_text']
    else:
        summary = mbart_summarizer(
            text, max_length=120, min_length=30, do_sample=False
        )[0]['summary_text']

    return {
        "document_language": lang,
        "summary": summary
    }
```

```python
doc_en = """The Kochi Metro project is focusing on expanding Phase 2 with new stations across the city.
The report highlights the importance of digital ticketing and passenger safety.
It also recommends installing more surveillance cameras at crowded junctions.
The expansion is expected to reduce traffic congestion significantly.
"""
print(hybrid_summarize(doc_en))
```
